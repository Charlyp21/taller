## 1 · La orden 🤵

Del menú de Open-Meteo pedimos el *endpoint* del archivo histórico. La orden lleva
**parámetros**: dónde (latitud/longitud), cuándo, y qué variables diarias queremos.
Fíjate: es un **GET** — solo leemos, no cambiamos nada en la cocina.

````python
import requests

URL = "https://archive-api.open-meteo.com/v1/archive"     # el endpoint del menú
PARAMS = {
    "latitude": 19.43, "longitude": -99.13,               # Ciudad de México
    "start_date": "2024-01-01", "end_date": "2025-12-31",
    "daily": ",".join([
        "temperature_2m_max", "temperature_2m_min", "temperature_2m_mean",
        "precipitation_sum", "windspeed_10m_max", "shortwave_radiation_sum",
    ]),
    "timezone": "America/Mexico_City",
}

r = requests.get(URL, params=PARAMS, timeout=30)
print("status_code:", r.status_code)   # 🔮 DEBES VER: 200 (todo bien). 404 = platillo inexistente; 429 = demasiadas órdenes
````

## 2 · El platillo 🍽️

La respuesta viene en **JSON** — que en Python es… ¡un diccionario! Ya sabes moverte ahí.
Explora antes de convertir: ¿qué llaves trae? ¿dónde están los datos de verdad?

````python
data = r.json()
print("llaves de la respuesta:", list(data.keys()))
print("llaves de data['daily']:", list(data["daily"].keys()))
print("la cocina confirma el punto:", data["latitude"], data["longitude"])
````

## 3 · De JSON a tabla

Un diccionario de listas paralelas es exactamente lo que `pandas` sabe convertir en tabla.

````python
import pandas as pd

df = pd.DataFrame(data["daily"]).rename(columns={"time": "fecha"})
print(df.shape)          # 🔮 DEBES VER: (731, 7) — dos años, día por día
df.head()
````

## 4 · Una probadita (el EDA formal llega la próxima clase)

````python
df["precipitation_sum"].describe()
````

## 5 · Guardar el platillo

El CSV se queda en tu repo: **la próxima clase lo metemos a una base de datos** y le

````python
df.to_csv("clima_cdmx.csv", index=False)
print("✅ clima_cdmx.csv guardado —", len(df), "días")
````

````python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib.colors import Normalize

URL = "https://archive-api.open-meteo.com/v1/archive"
PARAMS = {
    "latitude": 18.61472, "longitude": -99.18028,
    "start_date": "2024-01-01", "end_date": "2025-12-31",
    "daily": ",".join([
        "temperature_2m_max", "temperature_2m_min", "temperature_2m_mean",
        "precipitation_sum", "windspeed_10m_max", "shortwave_radiation_sum",
    ]),
    "timezone": "America/Mexico_City",
}

r = requests.get(URL, params=PARAMS, timeout=30)
data = r.json()
df = pd.DataFrame(data["daily"]).rename(columns={"time": "fecha"})

# --- Datos ---
y = df["temperature_2m_max"].to_numpy()
x = np.arange(len(y))

# --- Construir segmentos (cada par de puntos consecutivos) ---
points = np.array([x, y]).T.reshape(-1, 1, 2)
segments = np.concatenate([points[:-1], points[1:]], axis=1)

# Valor asociado a cada segmento: promedio de sus extremos
seg_values = (y[:-1] + y[1:]) / 2

# --- Colormap: azul (frío) -> rojo (caliente) ---
norm = Normalize(vmin=y.min(), vmax=y.max())
cmap = plt.get_cmap("coolwarm")   # o "RdBu_r", "turbo", etc.

lc = LineCollection(segments, cmap=cmap, norm=norm, linewidth=1.5)
lc.set_array(seg_values)

# --- Plot ---
fig, ax = plt.subplots(figsize=(10, 3))
ax.add_collection(lc)
ax.set_xlim(x.min(), x.max())
ax.set_ylim(y.min() - 1, y.max() + 1)
ax.set_xlabel("día (0 = 1 de enero de 2024)")
ax.set_title("Jojutla · temperatura máxima diaria (°C)")

# Barra de color
cbar = fig.colorbar(lc, ax=ax)
cbar.set_label("°C")

plt.tight_layout()
plt.show()
````

````python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.collections import LineCollection
from matplotlib.colors import LinearSegmentedColormap, Normalize

# ---------- 1. Descarga de datos ----------
URL = "https://archive-api.open-meteo.com/v1/archive"
PARAMS = {
    "latitude": 18.61472, "longitude": -99.18028,
    "start_date": "2024-01-01", "end_date": "2025-12-31",
    "daily": ",".join([
        "temperature_2m_max", "temperature_2m_min", "temperature_2m_mean",
        "precipitation_sum", "windspeed_10m_max", "shortwave_radiation_sum",
    ]),
    "timezone": "America/Mexico_City",
}

r = requests.get(URL, params=PARAMS, timeout=30)
data = r.json()
df = pd.DataFrame(data["daily"]).rename(columns={"time": "fecha"})

# Quitar días sin dato para no romper los segmentos
df = df.dropna(subset=["temperature_2m_max"]).reset_index(drop=True)

# ---------- 2. Preparar los datos ----------
y = df["temperature_2m_max"].to_numpy()
x = np.arange(len(y))

# Segmentos entre cada par de puntos consecutivos
points = np.array([x, y]).T.reshape(-1, 1, 2)
segments = np.concatenate([points[:-1], points[1:]], axis=1)

# Valor de temperatura asociado a cada segmento (promedio de sus extremos)
seg_values = (y[:-1] + y[1:]) / 2

# ---------- 3. Umbrales y colormap ----------
T_FRIO  = 15   # <= 15 °C  -> azul
T_CALOR = 25   # >= 25 °C  -> rojo

cmap = LinearSegmentedColormap.from_list(
    "frio_templado_calor",
    [(0.00, "#1f77b4"),   # azul (<= 15)
     (0.50, "#eeeeee"),   # blanco/gris (en 20, punto medio 15-25)
     (1.00, "#d62728")],  # rojo (>= 25)
)

# Normalización centrada en 15-25; con clip=True los valores
# fuera de ese rango se quedan con el color extremo
norm = Normalize(vmin=T_FRIO, vmax=T_CALOR, clip=True)

# ---------- 4. Plot ----------
fig, ax = plt.subplots(figsize=(10, 3))

lc = LineCollection(segments, cmap=cmap, norm=norm, linewidth=1.6)
lc.set_array(seg_values)
ax.add_collection(lc)

ax.set_xlim(x.min(), x.max())
ax.set_ylim(y.min() - 1, y.max() + 1)
ax.set_xlabel("día (0 = 1 de enero de 2024)")
ax.set_title("Jojutla · temperatura máxima diaria (°C)")

# Barra de color con marcas en los umbrales
cbar = fig.colorbar(lc, ax=ax, ticks=[T_FRIO, (T_FRIO + T_CALOR) / 2, T_CALOR])
cbar.ax.set_yticklabels([
    f"{T_FRIO} °C",
    f"{(T_FRIO + T_CALOR) / 2:.0f} °C",
    f"{T_CALOR} °C",
])

plt.tight_layout()
plt.show()
````
